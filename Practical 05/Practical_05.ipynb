```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow
```

Question 01

```python
image = cv2.imread('tiger.jpg',cv2.IMREAD_GRAYSCALE)

#Calculate the histogram of the image
hist = cv2.calcHist([image],[0],None,[256],[0,256])

#Plot the original image and the histogram.
plt.subplot(2,2,1)
plt.imshow(image,cmap='gray')
plt.title('Original Image')
plt.subplot(2,2,2)
plt.plot(hist)
plt.title('Histogram')
plt.show()
```

```python
#histogram equalization
equ = cv2.equalizeHist(image)

#display original image and histogram equalized image
plt.subplot(2,2,1)
plt.imshow(image,cmap='gray')
plt.title('Original Image')
plt.subplot(2,2,2)
plt.imshow(equ,cmap='gray')
plt.title('Histogram Equalized Image')


```

```python
#Calculate the histogram of the image.
hist_equ = cv2.calcHist([equ],[0],None,[256],[0,256])

#display the histogram of equalized histogram image

plt.plot(hist_equ)
plt.title('Histogram of Equalized Image')
plt.show()
```

Question 3


```python
image_lc= cv2.imread("lcimg.jpg",cv2.COLOR_RGB2BGR)
cv2_imshow(image_lc)
```

```python
#seperate color channels
B=image_lc[:,:,0] #blue layer
G=image_lc[:,:,1] #green layer
R=image_lc[:,:,2] #red layer
```

```python
#histogram of color channels
B_histo=cv2.calcHist([B],[0],None,[256],[0,256])
G_histo=cv2.calcHist([G],[0],None,[256],[0,256])
R_histo=cv2.calcHist([R],[0],None,[256],[0,256])
```

```python
#display histograms
plt.subplot(2,2,1)
plt.plot(B_histo,'b')

plt.subplot(2,2,2)
plt.plot(G_histo,'g')

plt.subplot(2,2,3)
plt.plot(R_histo,'r')

plt.show()
```

```python
#equalize color channels
B_eq=cv2.equalizeHist(B)
G_eq=cv2.equalizeHist(G)
R_eq=cv2.equalizeHist(R)
```

```python
#equilized histograms
B_histo_eq=cv2.calcHist([B_eq],[0],None,[256],[0,256])
G_histo_eq=cv2.calcHist([G_eq],[0],None,[256],[0,256])
R_histo_eq=cv2.calcHist([R_eq],[0],None,[256],[0,256])
```

```python
#show equalized histograms
plt.subplot(2,2,1)
plt.plot(B_histo_eq,'b')

plt.subplot(2,2,2)
plt.plot(G_histo_eq,'g')

plt.subplot(2,2,3)
plt.plot(R_histo_eq,'r')
```

```python
#merge the channels and create new image
image_eq=cv2.merge((B_eq,G_eq,R_eq))
cv2_imshow(image_lc)
cv2_imshow(image_eq)
```
