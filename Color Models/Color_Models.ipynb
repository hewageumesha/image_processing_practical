```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('Color_image.jpg') # by default loaded as RGB
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB) # convert BGR to RGB
image_hsv = cv2.cvtColor(image_rgb, cv2.COLOR_BGR2HSV) # convert RGB to HSV

plt.subplot(1, 3, 1)
plt.imshow(image_rgb)
plt.title("RGB Image")

plt.subplot(1, 3, 2)
plt.imshow(image_hsv)
plt.title("HSV Image")

plt.subplot(1, 3, 3)
plt.imshow(image)
plt.title("Original Image")

plt.show()
```

```python
# formula to convert RGB to CNY
def rgb_to_cmy(r, g, b):

  # RGB values are divided by 255
  # to bring them between 0 to 1
  c = 1 - r / 255
  m = 1 - g / 255
  y = 1 - b / 255

  return (c, m, y)

# sample RGB values
r = 0
g = 169
b = 86

# print the result
print(rgb_to_cmy(r, g, b))

# c = 1.0 -> full cyan ink
# m = 0.337 -> small amount of magenta ink
# y = 0.663 -> medium amount of yellow ink
```
