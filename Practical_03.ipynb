Arithmetic Operations (Addition & Subtraction)

```python
#Addition
# Read images
image1 = cv2.imread('image1.webp')
image2 = cv2.imread('image2.jpg')

# Resize image2 to match image1
image2_resized = cv2.resize(image2, (image1.shape[1], image1.shape[0]))

# Perform addition
add = cv2.add(image1, image2_resized)

# Save result
cv2.imwrite('addition.jpg', add)

# Display
cv2_imshow(cv2.cvtColor(add, cv2.COLOR_BGR2RGB))
```

```python
#Subtraction
# Read images
image1 = cv2.imread('image1.webp')
image2 = cv2.imread('image2.jpg')

# Resize image2 to match image1
image2 = cv2.resize(image2, (image1.shape[1], image1.shape[0]))

# Subtraction
sub = cv2.subtract(image1, image2)

# Save output
cv2.imwrite('subtraction.jpg', sub)

# Display
cv2_imshow(cv2.cvtColor(sub, cv2.COLOR_BGR2RGB))
```

Image Blending Operation

```python
image1 = cv2.imread('blend1.webp')
image2 = cv2.imread('blend2.jpg')

# Resize
image2 = cv2.resize(image2, (image1.shape[1], image1.shape[0]))

# Blending parameters
alpha = 0.7
beta = 0.3
gamma = 0

# Blend images
blended = cv2.addWeighted(image1, alpha, image2, beta, gamma)

# Save output
cv2.imwrite('blended.jpg', blended)

# Display
cv2_imshow(cv2.cvtColor(blended, cv2.COLOR_BGR2RGB))
```

Bitwise Operations

```python
#14a & 14b. Rectangle Image (300×300)
import numpy as np

# Create black image
rect = np.zeros((300, 300), dtype=np.uint8)

# Draw white filled rectangle
cv2.rectangle(rect, (25,25), (275,275), 255, -1)

cv2_imshow(rect)
```

```python
#14c & 14d. Circle Image (300×300)
# Create black image
circle = np.zeros((300, 300), dtype=np.uint8)

# Draw white filled circle
cv2.circle(circle, (150,150), 150, 255, -1)

cv2_imshow(circle)
```

```python
#14e. Bitwise Operations (AND, OR, XOR, NOT)
bit_and = cv2.bitwise_and(rect, circle)
bit_or  = cv2.bitwise_or(rect, circle)
bit_xor = cv2.bitwise_xor(rect, circle)
bit_not = cv2.bitwise_not(rect)

# Display results
cv2_imshow(bit_and)
cv2_imshow(bit_or)
cv2_imshow(bit_xor)
cv2_imshow(bit_not)
```
