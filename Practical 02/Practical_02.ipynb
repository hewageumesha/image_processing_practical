Import Library

```python
import cv2
from google.colab.patches import cv2_imshow
```

Read and Display the image

```python
#read image file
image = cv2.imread('image_1.jpg')

#show image
cv2_imshow(image)
```

Convert to grayscale image and display

```python
#convert the grayscale
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

#save the grayscale image
cv2.imwrite('gray_image.jpg', gray_image)

#display image
cv2_imshow(gray_image)
```

Resize image

```python
#resized the image
resized_image = cv2.resize(image, (600,400))

#save the resized image
cv2.imwrite('resized_image.jpg', resized_image)

#display image
cv2_imshow(resized_image)
```

Rotate Image

```python
#get the image's height and width
(h, w) = image.shape[:2]

#define the center of the image(using // round the result down to the nearest whole number)
center = (w // 2, h // 2)

#creates a rotation matrix that rotates the image 90 degrees counterclockwise around its center.
#1.0 -> keep original size
matrix = cv2.getRotationMatrix2D(center , 180, 1.0)

#perform the rotation
rotate_image = cv2.warpAffine(image, matrix, (w,h))

#save the rotated image
cv2.imwrite('rotated_image.jpg', rotate_image)

#show image
cv2_imshow(rotate_image)
```

Gaussian Blur

```python
#blurred image
blurred_image = cv2.GaussianBlur(image, (21,21), 0)

#save blurred image
cv2.imwrite('blurred_image.jpg', blurred_image)

#display blurred image
cv2_imshow(blurred_image)
```

Canny edge detector

```python
#canny edge detector image
edge_image = cv2.Canny(gray_image, 100, 200)

#save canny edge image
cv2.imwrite('canny_edge_image.jpg', edge_image)

#display edge image
cv2_imshow(edge_image)
```

Drawing shape and adding text

```python
#load image
image = cv2.imread('image_1.jpg')

#draw rectangle
cv2.rectangle(image, (50,50), (200,200), (255,0,0), 10)

#add text to the image
cv2.putText(image, 'TG/2021/1067 N.U.M.Hewage', (60,40), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 0), 2)

#save the output image
cv2.imread('output_image.jpg', image)

#display image
cv2_imshow(image)
```

Threshould image

```python
# Load image
image = cv2.imread('image_1.jpg')

# Apply threshold
_, thresh_image = cv2.threshold(image, 130, 255, cv2.THRESH_BINARY)

# Save the output image
cv2.imwrite('thres_image.jpg', thresh_image)

# Load threshold image
image = cv2.imread('thres_image.jpg', cv2.IMREAD_GRAYSCALE)

# Display image
cv2_imshow(image)
```
